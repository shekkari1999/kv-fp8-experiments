```python
## Check config

from transformers import AutoConfig

config = AutoConfig.from_pretrained("Qwen/Qwen3-1.7B")
config
```

```python
layers = config.num_hidden_layers
hidden_size = config.hidden_size
attention_heads = config.num_attention_heads
kv_heads = config.num_key_value_heads

head_dimension = hidden_size // attention_heads
KV_BF16  = 2 * layers * kv_heads * head_dimension * 2 * 4096
KV_FP8  = 2 * layers * kv_heads * head_dimension * 1 * 4096
print(f'The KV cache for 4096 tokens in BF16 is {KV_BF16/ 1024**2:.1f} MB')
print(f'The KV cache for 4096 tokens in FP8 is {KV_FP8/ 1024**2:.1f} MB')

```
